```python
documents = [
    "cat dog",
    "cat mouse",
    "dog mouse",
    "cat cat dog"
]
```

```python
vocab = ' '.join([doc for doc in documents])
vocab = set(vocab.split())
# del words
```

```python
def compute_tf(docs:str)->dict:
    # words = ' '.join([doc for doc in docs]).split()
    words = docs.split()
    total_words = len(words)
    # vocab = set(words.split())

    tf:map[str, float] = {}
    for word in words:
        tf[word] = tf.get(word, 0) + 1
    for word in tf:
        tf[word] /= total_words
        
    # print(tf)
    return tf
```

```python
tf = compute_tf(documents[3])

assert tf["cat"] == 2/3
assert tf["dog"] == 1/3
```

```python
import math
```

```python
def compute_idf(docs:list)->dict:
    idf = {}
    words = ' '.join([doc for doc in docs])
    vocab = set(words.split())
    
    for sentence in docs:
        words = set(sentence.split())
        for word in words:
            idf[word] = idf.get(word, 0) + 1

    n = len(docs)
    for key in idf:
        idf[key] = math.log(n/idf[key])

    # print(idf)
    return idf
```

```python
idf = compute_idf(documents)

assert math.isclose(idf["cat"], math.log(4/3), rel_tol=1e-5)
assert math.isclose(idf["dog"],math.log(4/3), rel_tol=1e-5)
assert math.isclose(idf["mouse"],math.log(2),rel_tol=1e-5)
```

```python
def compute_tfidf(docs: list):
    idf = compute_idf(docs)
    tfidf = []

    for doc in docs:
        tf = compute_tf(doc)
        vector = {}

        for word in idf:
            vector[word] = tf.get(word, 0) * idf[word]

        tfidf.append(vector)

    return tfidf
```

```python
tfidf = compute_tfidf(documents)

expected = 0.5 * math.log(4/3)

assert math.isclose(tfidf[0]["cat"],expected,rel_tol=1e-5)
assert math.isclose(tfidf[0]["dog"],expected,rel_tol=1e-5)
```
