```python
sentence="The blue kite danced wildly in the strong afternoon wind before it finally caught on a high branch.Apple bought a startup in the UK for $1 billion on Tuesday"
```

```python
import nltk
nltk.download('maxent_ne_chunker_tab')
```

```python
words=[word for word in sentence.split()]
```

```python
tag_elements=nltk.pos_tag(words)
```

```python
nltk.download('words')
```

```python
nltk.ne_chunk(tag_elements).draw()
```
