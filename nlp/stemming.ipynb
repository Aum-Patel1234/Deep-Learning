```python
# stemming is a process of reducing a word to its suffixes and affixes or to 
# the root of the word also known as lemma.
```

```python
import nltk
```

```python
words = ["running", "runs", "ran", "easily", "fairly"]
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
for w in words:
    print(w,' - ', stemmer.stem(w))
```

```python
# second type of stemmer
```

```python
from nltk.stem import RegexpStemmer
```

```python
regex_stemmer = RegexpStemmer('ing$|s$|e$|able$')
```

```python
for w in words:
    print(w,' - ', regex_stemmer.stem(w))
```

```python
stemmer.stem("ingeating")
```

```python
regex_stemmer = RegexpStemmer('ing|s$|e$|able$') # remove dollar after ing so its checks `ing` throughout
regex_stemmer.stem("ingeating")
```

```python
# snowball stemmer
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowball_stemmer=SnowballStemmer('english')
```

```python
for w in words:
    print(w,' - ', snowball_stemmer.stem(w))
```

```python
# BELOW is where Porterstemmer and SnowballStemmer gives diff ans
```

```python
stemmer.stem('fairly'), stemmer.stem('sportingly')
```

```python
snowball_stemmer.stem('fairly'), snowball_stemmer.stem('sportingly')
```
