```python
import nltk
```

```python
words=["running", "eating", "working", "programming"]
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
nltk.download('wordnet')
```

```python
for w in words:
    print(w, '-', lemmatizer.lemmatize(w, pos='v'))
```

```python
for w in words:
    print(w, '-', lemmatizer.lemmatize(w, pos='r'))
```

```python
for w in words:
    print(w, '-', lemmatizer.lemmatize(w, pos='n'))
```
