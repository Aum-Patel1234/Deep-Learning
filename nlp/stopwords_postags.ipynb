```python
para="""
Artificial Intelligence (AI) is transforming the modern world at an unprecedented pace. Dr. Elena Vance, a leading computer scientist, told reporters on Tuesday that 'intelligent machines are no longer a futuristic dream.' Currently, her team in San Francisco is developing algorithms that process natural language with 98.5% accuracy.
"""
```

```python
print(para)
```

```python
import nltk
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
len(stopwords.words()), len(stopwords.words('french'))
```

```python
from nltk.stem import PorterStemmer, SnowballStemmer
from nltk.tokenize import sent_tokenize
```

```python
stemmer=PorterStemmer()
```

```python
sentences=sent_tokenize(para)
```

```python
for i, sentence in enumerate(sentences):
    words = nltk.word_tokenize(sentence)
    words = [stemmer.stem(word) for word in words if word not in set(stopwords.words())]
    sentences[i] = ' '.join(words)
```

```python
sentences
```

```python
# snowball
sentences=sent_tokenize(para)
```

```python
stemmer=SnowballStemmer('english')
for i, sentence in enumerate(sentences):
    words = nltk.word_tokenize(sentence)
    words = [stemmer.stem(word) for word in words if word not in set(stopwords.words())]
    sentences[i] = ' '.join(words)
```

```python
sentences
```

```python
sentences=sent_tokenize(para)
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
for i, sentence in enumerate(sentences):
    words = nltk.word_tokenize(sentence)
    words = [lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words())]
    sentences[i] = ' '.join(words)
```

```python
sentences
```

```python
sentences=sent_tokenize(para)
for i, sentence in enumerate(sentences):
    words = nltk.word_tokenize(sentence)
    words = [lemmatizer.lemmatize(word, 'v') for word in words if word not in set(stopwords.words())]
    sentences[i] = ' '.join(words)
```
